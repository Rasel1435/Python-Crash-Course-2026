## Conditional Statement

-   if
-   else
-   elif
-   nested if-else

### if-else (Single condition)

```python
pizza_size = input("Enter pizza size (Large/Regular): ")

bill = 0

if pizza_size == "L" or pizza_size == "l":
    bill = 300
    print(f"Large Pizza selected. Base Price: {bill} TK.")
else:
    bill = 150
    print(f"Regular Pizza selected. Base Price: {bill} TK.")

print(f"Total Bill: {bill} TK.")
```

### elif statement (Multiple condition)

```python
pizza_size = input("Enter pizza size (Large/Medium/Small): ")
bill = 0

if pizza_size == "Large" or pizza_size == "large":
    bill = 900
elif pizza_size == "Medium" or pizza_size == "medium":
    bill = 700
elif pizza_size == "Small" or pizza_size == "small":
    bill = 500
else:
    print("Invalid size! Defaulting to Small Pizza.")
    bill = 100

print(f"Current Bill: {bill} TK.")
```

### Nested if-else (Multiple conditions with in another condition)

```python
pizza_size = input("Enter pizza size (Large/Medium/Small): ")
add_cheese = input("Do you want Extra Cheese? (Yes/No): ")
bill = 0

if pizza_size == "Large":
    bill = 300
    if add_cheese == "Yes":
        bill = bill + 50 
        print("Extra Cheese added for Large Pizza (+50 TK).")

elif pizza_size == "Medium":
    bill = 200
    if add_cheese == "Yes":
        bill = bill + 30  
        print("Extra Cheese added for Medium Pizza (+30 TK).")

else:
    bill = 100
    if add_cheese == "Yes":
        bill = bill + 20 
        print("Extra Cheese added for Small Pizza (+20 TK).")

print(f"Final Bill after Cheese Check: {bill} TK.")
```

### E-Commerce Pizza Express Billing System

```python
print("====================================")
print("🍕 WELCOME TO PIZZA EXPRESS 2026 🍕")
print("====================================")

size = input("Choose Size (Large/Medium/Small): ")
bill = 0

if size == "Large":
    bill = 300
elif size == "Medium":
    bill = 200
elif size == "Small":
    bill = 100
else:
    print("Wrong choice! Automatically selected Small.")
    size = "Small"
    bill = 100

cheese = input("Want Extra Cheese? (Yes/No): ")
if cheese == "Yes":
    if size == "Large":
        bill += 50
    elif size == "Medium":
        bill += 30
    else:
        bill += 20

drinks = input("Want a Cold Drink? (Yes/No): ")
if drinks == "Yes":
    bill += 40
    print("Cold Drink Added (+40 TK).")

print(f"\nSubtotal: {bill} TK.")

promo_code = input("\nEnter Promo Code (If any, else press Enter): ")

if promo_code == "PIZZA20":
    print("🎉 Promo Code Applied Successfully! You got 20 TK Discount.")
    bill = bill - 20
else:
    if promo_code != "":
        print("❌ Invalid Promo Code! No discount applied.")

print("\n====================================")
print("             CASH MEMO              ")
print("====================================")
print(f"Pizza Size : {size}")
print(f"Extra Cheese: {cheese}")
print(f"Cold Drinks : {drinks}")
print("------------------------------------")
print(f"TOTAL PAYABLE: {bill} TK")
print("====================================")
print("Thank you for ordering with us! ❤️")
```
