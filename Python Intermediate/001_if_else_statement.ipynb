## Conditional Statement

-   if
-   else
-   elif
-   nested if-else

### if-else (Single condition)

```python
pizza_size = input('Enter pizza size (Large/Regular): ').strip().lower()

bill = 0

if pizza_size == "l":
    bill = 900
    print(f"Large pizza selected. Base Price: {bill} TK.")
else:
    bill = 150
    print(f"Regular Pizza selected. Base price: {bill} TK.")

print(f"Total Bill: {bill} TK")
```

### elif statement (Multiple condition)

```python
pizza_size = input("Enter pizza size (Large/Medium/Small): ").strip().lower()
bill = 0

if pizza_size == "large" or pizza_size == "l":
    bill = 900
elif pizza_size == "medium" or pizza_size == "m":
    bill = 700
elif pizza_size == "small" or pizza_size == "s":
    bill = 500
else:
    print("Invalid size! Defaulting to Small Pizza.")
    bill = 100

print(f"Current Bill: {bill} TK.")
```

### Nested if-else (Multiple conditions with in another condition)

```python
pizza_size = input("Enter pizza size (Large/Medium/Small): ").strip().lower()
add_cheese = input("Do you want Extra Cheese? (Yes/No): ").strip().lower()
bill = 0

if pizza_size == "large":
    bill = 300
    if add_cheese == "yes":
        bill = bill + 50 
        print("Extra Cheese added for Large Pizza (+50 TK).")

elif pizza_size == "medium":
    bill = 200
    if add_cheese == "yes":
        bill = bill + 30  
        print("Extra Cheese added for Medium Pizza (+30 TK).")

else:
    bill = 100
    if add_cheese == "yes":
        bill = bill + 20 
        print("Extra Cheese added for Small Pizza (+20 TK).")

print(f"Final Bill after Cheese Check: {bill} TK.")
```

### E-Commerce Pizza Express Billing System

```python
print("====================================")
print("🍕 WELCOME TO PIZZA EXPRESS 2026 🍕")
print("====================================")

print("\n--- TODAY'S PIZZA MENU ---")
print("1. Large Pizza  : 900 TK")
print("2. Medium Pizza : 700 TK")
print("3. Small Pizza  : 500 TK")
print("---------------------------\n")

size = input("Choose Size (Large/Medium/Small): ").strip().lower()
bill = 0

if size == "large":
    bill = 900
    print(f"You Selected Large One {bill} TK.")
elif size == "medium":
    bill = 700
    print(f"You Selected Medium One {bill} TK.")
elif size == "small":
    bill = 500
    print(f"You Selected Small One {bill} TK.")
else:
    print("Sorry! You didn't choose a valid size. Program closing... Good luck next time! 😊")


if size == "large" or size == "medium" or size == "small":
    
    cheese = input("Want Extra Cheese? (Yes/No): ").strip().lower()
    if cheese == "yes":
        if size == "large":
            bill += 50
            print("Cheese added (+50 TK).")
        elif size == "medium":
            bill += 30
            print("Cheese added (+30 TK).")
        elif size == "small":
            bill += 20
            print("Cheese added (+20 TK).")

    drinks = input("Want a Cold Drink? (Yes/No): ").strip().lower()
    if drinks == "yes":
        bill += 40
        print("Cold Drink Added (+40 TK).")

    print(f"\nSubtotal: {bill} TK.")


    promo_code = input("\nEnter Promo Code (If any, else press Enter): ")

    if promo_code == "PIZZA20":
        print("🎉 Promo Code Applied Successfully! You got 20 TK Discount.")
        bill = bill - 20
    else:
        if promo_code != "":
            print("Invalid Promo Code! No discount applied.")


    print("\n====================================")
    print("             CASH MEMO              ")
    print("====================================")
    print(f"Pizza Size : {size.capitalize()}")
    print(f"Extra Cheese: {cheese.capitalize()}")
    print(f"Cold Drinks : {drinks.capitalize()}")
    print("------------------------------------")
    print(f"TOTAL PAYABLE: {bill} TK")
    print("====================================")
    print("Thank you for ordering with us! ❤️")
```
