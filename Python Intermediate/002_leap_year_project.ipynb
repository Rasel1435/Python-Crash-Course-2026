## Leap Year
লিপ ইয়ার (Leap Year) বা অধিবর্ষ মানে হলো এমন একটি বছর, যে বছরে সাধারণ ৩৬৫ দিনের বদলে ক্যালেন্ডারে ১টি দিন বেশি থাকে, অর্থাৎ বছরটি হয় ৩৬৬ দিনের। আর এই বাড়তি দিনটি যোগ করা হয় বছরের সবচেয়ে ছোট মাস ফেব্রুয়ারির সাথে (ফেব্রুয়ারি তখন ২৮ দিনের বদলে ২৯ দিনের হয়)।

কিন্তু মনে প্রশ্ন আসতে পারে— "আসোলে বা বাস্তবে কেন ৪ বছর পর পর এই ১টি দিন হুট করে বেড়ে যায়?" এর পেছনে রয়েছে সম্পূর্ণ একটি মহাজাগতিক ও বৈজ্ঞানিক কারণ। নিচে একদম সহজ ভাষায় আসল রহস্যটি বুঝিয়ে দেওয়া হলো:

#### 🌍 আসল বৈজ্ঞানিক কারণ (The Scientific Mystery)
আমরা ছোটবেলা থেকে পড়ে আসছি পৃথিবী সূর্যের চারদিকে একবার ঘুরে আসতে (এক সৌর বছর) ঠিক ৩৬৫ দিন সময় নেয়। কিন্তু এটি কিন্তু একদম নিখুঁত হিসাব নয়!

বাস্তবে, সূর্যের চারদিকে একবার ঘুরে আসতে পৃথিবীর সময় লাগে ৩৬৫ দিন, ৫ ঘণ্টা, ৪৮ মিনিট এবং ৪৬ সেকেন্ড (প্রায় ৩৬৫ দিন এবং ৬ ঘণ্টার মতো)।

এখন চিন্তা করুন, আমরা যদি প্রতি বছর ক্যালেন্ডারে শুধু ৩৬৫ দিন ধরি, তাহলে প্রতি বছর এই প্রায় ৬ ঘণ্টা সময় অলক্ষিত বা বাদ থেকে যাচ্ছে। এভাবে হিসাব করলে:

-   ১ম বছর শেষে ঘাটতি: ৬ ঘণ্টা
-   ২য় বছর শেষে ঘাটতি: ১২ ঘণ্টা
-   ৩য় বছর শেষে ঘাটতি: ১৮ ঘণ্টা
-   ৪র্থ বছর শেষে ঘাটতি: ২৪ ঘণ্টা (অর্থাৎ পুরো ১টি দিন!)

আমরা যদি এই বাড়তি ২৪ ঘণ্টাকে বা ১টি দিনকে ক্যালেন্ডারে যোগ না করি, তবে প্রতি ৪ বছর পর পর আমাদের ক্যালেন্ডার প্রকৃতির ঋতুচক্র থেকে ১ দিন পিছিয়ে পড়বে। এভাবে ১০০ বছর পর ক্যালেন্ডার প্রায় ২৪ দিন পিছিয়ে যাবে! তখন দেখা যাবে বাংলাদেশে জুন মাসের বর্ষাকাল চলে গেছে জুলাই বা আগস্ট মাসে, আর পৌষ মাসের শীত চলে এসেছে মাঘ বা ফাল্গুন মাসে।

ঋতু ও প্রকৃতির সাথে ক্যালেন্ডারের এই মিল বজায় রাখার জন্যই প্রতি ৪ বছর পর পর এই জমে থাকা ২৪ ঘণ্টা বা ১টি দিনকে ফেব্রুয়ারির ২৯ তারিখ হিসেবে যোগ করে দেওয়া হয়। একেই বলে লিপ ইয়ার!

#### 🚦 শতাব্দীর বছরগুলোর (Centennial Years) বিশেষ টুইস্ট
অনেকে মনে করেন প্রতি ৪ দিয়ে ভাগ গেলেই বুঝি লিপ ইয়ার হয়। কিন্তু এর ভেতরে আরেকটি গভীর গাণিতিক নিয়ম আছে:

আমরা যে প্রতি বছর ৬ ঘণ্টা করে বাড়তি ধরছি, সেটি কিন্তু নিখুঁত ৬ ঘণ্টা নয় (৬ ঘণ্টা থেকে ১১ মিনিট ১৪ সেকেন্ড কম)। এই সামান্য ১১ মিনিট প্রতি ৪ বছরে জমতে জমতে ৪০০ বছর পর প্রায় ৩ দিন বেশি হয়ে যায়!

এই ৩ দিনের অতিরিক্ত হিসাব কমাতে বিজ্ঞানীরা আরেকটি নিয়ম বানিয়েছেন:

"যেসব বছরের শেষে দুটি শূন্য থাকে (শতাব্দীর বছর, যেমন: ১৯০০, ২১০০, ২২০০), সেগুলোকে শুধু ৪ দিয়ে ভাগ গেলেই হবে না, সেগুলোকে ৪০০ দিয়েও ভাগ যেতে হবে।"

২০০০ সাল: এটি ৪০০ দিয়ে ভাগ যায়, তাই এটি লিপ ইয়ার ছিল।

২১০০ সাল (আমাদের পরবর্তী শতাব্দীর বছর): এটি ৪ দিয়ে ভাগ গেলেও ৪০০ দিয়ে ভাগ যায় না, তাই এটি কিন্তু লিপ ইয়ার হবে না (সে বছর ফেব্রুয়ারি ২৮ দিনেই শেষ হবে)।

### শুধুমাত্র ৪ দিয়ে ভাগ চেক করা (If-Else)
একেবারে শুরুতে আমরা শুধু বেসিক নিয়মটি চেক করব। পাইথনে নিঃশেষে বিভাজ্য চেক করার জন্য মডুলাস বা ভাগশেষ অপারেটর (%) ব্যবহার করা হয়।

```python
year = int(input("Enter a year to check: "))

if year % 4 == 0:
    print(f"{year} might be a Leap Year!")
else:
    print(f"{year} is NOT a Leap Year.")
```

আপনি যদি এখানে ২১০০ ইনপুট দেন, কোড বলবে এটি লিপ ইয়ার। কিন্তু বাস্তবে ২১০০ সাল লিপ ইয়ার নয়! এই সমস্যা সমাধানের জন্য আমাদের ২য় ধাপে যেতে হবে।

### শতাব্দীর বছর হ্যান্ডেল করা (Nested If-Else)
এবার আমরা এক্সাম্পল ১-এর কোডটিকে বড় করব এবং ভেতরে ১০০ ও ৪০০ এর শর্ত দুটি নেস্টেড (একটার ভেতর আরেকটা) আকারে ঢুকিয়ে দেব।

```python
year = int(input("Enter a year to check: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a Leap Year! (Centennial & Divisible by 400)")
        else:
            print(f"{year} is NOT a Leap Year. (Centennial but not divisible by 400)")
    else:
        print(f"{year} is a Leap Year!")
else:
    print(f"{year} is NOT a Leap Year.")
```

### Smart Calendar & Event Planner 2026

```python
print("=========================================")
print("SMART CALENDAR & EVENT PLANNER 2026")
print("=========================================")

year = int(input("Enter Year (e.g., 2024, 2100): "))

if year <= 0:
    print("Invalid Year! Program closing... Good luck next time!")
else:
    is_leap_year = False

    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                is_leap_year = True
            else:
                is_leap_year = False
        else:
            is_leap_year = True
    else:
        is_leap_year = False

    print("\n--- MONTHS LIST ---")
    print("January, February, March, April, May, June, July, August, September, October, November, December")
    print("-------------------")
    
    month = input("Enter Month Name: ").strip().lower()
    days = 0
    valid_month = True

    if month in ["january", "march", "may", "july", "august", "october", "december"]:
        days = 31
    elif month in ["april", "june", "september", "november"]:
        days = 30
    elif month == "february":
        if is_leap_year == True:
            days = 29
        else:
            days = 28
    else:
        print("Invalid Month Name! Cannot process calendar.")
        valid_month = False

    if valid_month == True:
        print("\n=========================================")
        print("             CALENDAR REPORT             ")
        print("=========================================")
        print(f"Target Year  : {year}")
        print(f"Leap Year?   : {'YES ' if is_leap_year else 'NO'}")
        print(f"Selected Month: {month.capitalize()}")
        print("-----------------------------------------")
        print(f"TOTAL DAYS IN THIS MONTH: {days} DAYS")
        print("=========================================")
        
        if month == "february" and is_leap_year == True:
            print("Fun Fact: This February has an extra day because of the Leap Year cycle!")
        print("Thank you for using Smart Calendar!")
```
