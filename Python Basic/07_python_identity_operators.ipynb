## Python Identity Operators

Python-এ Identity Operators মূলত দুটি অবজেক্টের মেমোরি লোকেশন (Memory Location) বা আইডেন্টিটি (Identity) তুলনা করার জন্য ব্যবহার করা হয়। সহজ কথায়, দুটি ভেরিয়েবল একই মেমোরি অ্যাড্রেসে থাকা অবজেক্টকে নির্দেশ করছে কিনা, তা পরীক্ষা করার জন্য এটি ব্যবহার করা হয়।

Python-এ দুটি Identity Operator রয়েছে:
- `is`
- `is not`


`is` Operator
যদি দুটি ভেরিয়েবল একই অবজেক্টকে নির্দেশ করে (অর্থাৎ তাদের মেমোরি লোকেশন এক হয়), তবে এটি `True` রিটার্ন করে। অন্যথায় `False` রিটার্ন করে।

`is not` Operator
এটি `is` অপারেটরের উল্টো। যদি দুটি ভেরিয়েবল আলাদা আলাদা অবজেক্টকে নির্দেশ করে (অর্থাৎ মেমোরি লোকেশন এক না হয়), তবে এটি `True` রিটার্ন করে। আর মেমোরি লোকেশন এক হলে `False` রিটার্ন করে।

### How it works
পাইথনে ভেরিয়েবলগুলো ডেটা সরাসরি নিজের মধ্যে রাখে না, বরং ডেটার মেমোরি লোকেশনকে নির্দেশ করে।

- `==` অপারেটর দিয়ে আমরা চেক করি দুটি ভেরিয়েবলের Value (মান) সমান কিনা।

- `is` অপারেটর দিয়ে আমরা চেক করি ভেরিয়েবল দুটির Memory Address (id) এক কিনা।

পাইথনে কোনো অবজেক্টের মেমোরি অ্যাড্রেস দেখার জন্য id() ফাংশন ব্যবহার করা হয়।

### Example

```python
# -5 to 256

a = 20
b = 20

print(id(a))
print(id(b))

print(a is b)
```

```python
a = 257
b = 257

print(id(a))
print(id(b))

print(a is b)
```

### মূল কারণ (Integer Caching):
পাইথনে -5 থেকে 256 পর্যন্ত সংখ্যার জন্য মেমোরি অপ্টিমাইজেশনের একটা নিয়ম আছে।

- পাইথন প্রোগ্রাম চালানোর সময় মেমোরি বাঁচানোর জন্য এবং পারফরম্যান্স ফাস্ট করার জন্য আগে থেকেই -5 থেকে 256 পর্যন্ত সংখ্যাগুলোকে মেমোরিতে ফিক্সড বা ক্যাশ (Cache) করে রেখে দেয়।

- আপনি যখনই আপনার কোডে a = 5 লিখবেন, পাইথন নতুন মেমোরি তৈরি না করে আগে থেকে ক্যাশ করে রাখা 5 এর অ্যাড্রেসটি a-কে দিয়ে দেয়। একইভাবে যখন b = 5 লিখবেন, পাইথন আবার সেই একই অ্যাড্রেসটি b-কেও দিয়ে দেয়।

- যেহেতু a এবং b দুটো ভেরিয়েবলই মেমোরির একই স্থান বা একই অবজেক্টকে নির্দেশ করছে, তাই is অপারেটর এখানে True রিটার্ন করেছে।
