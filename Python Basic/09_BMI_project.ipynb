## BMI Project
BMI হলো একজন মানুষের শরীরের ওজন (Weight) এবং উচ্চতার (Height) অনুপাতের ওপর ভিত্তি করে তৈরি একটি আন্তর্জাতিক গাণিতিক হিসাব বা পরিমাপক। এর মাধ্যমে খুব সহজে নির্ধারণ করা যায় যে একজন মানুষের ওজন তার উচ্চতা অনুযায়ী ঠিক আছে নাকি বেশি বা কম।

সহজ কথায়, একজন মানুষ কি সুস্থ, নাকি অতিরিক্ত চিকন, নাকি অতিরিক্ত মোটা— তা এই ইনডেক্স বা সূচকের মাধ্যমে বৈজ্ঞানিক উপায়ে বের করা হয়।

Take Input from the user for weight and height, and find the BMI Value

BMI = weight/(height)**2

আন্তর্জাতিক পরিমাপ অনুযায়ী, ১ ইঞ্চি = ০.০২৫৪ মিটার।

### আন্তর্জাতিক BMI চার্ট (BMI Categories)
| BMI এর মান (Range) | শরীরের অবস্থা (Category) |
|-----------|----------|
| ১৮.৫ এর কম (< 18.5) | Underweight (ওজন প্রয়োজনের চেয়ে কম) |
| ১৮.৫ থেকে ২৪.৯ (18.5 - 24.9) | Normal Weight (একদম সঠিক ও সুস্থ ওজন) |
| ২৫ থেকে ২৯.৯ (25 - 29.9) | Overweight (ওজন প্রয়োজনের চেয়ে বেশি) |
| ৩০ বা তার বেশি ($\ge 30$) | Obese (স্থূলতা বা অতিরিক্ত মোটা) |
