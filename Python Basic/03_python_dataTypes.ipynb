## Data Types in Python

### Numeric Types
Numeric types in Python include integers, floating-point numbers, and complex numbers. They are used to represent numerical values and perform mathematical operations.

-   <strong>Integers</strong>: are whole numbers, positive or negative, without a decimal point. They can be of any length, as long as they fit within the memory of the computer.

-   <strong>Floating-point numbers</strong>: are numbers that have a decimal point. They can represent both positive and negative values, and they can also represent very large or very small numbers using scientific notation.
-   <strong>Complex numbers</strong>: are numbers that have a real part and an imaginary part. They are represented in the form `a + bj`, where `a` is the real part, `b` is the imaginary part, and `j` is the imaginary unit.

```python
a = 10
b = 3.14
c = 2 + 3j

print(a)
print(b)
print(c)
```

### Type Function
The `type()` function is used to determine the data type of a variable or value. It returns the type of the object passed to it.

```python
print(type(a))
print(type(b))
print(type(c))
```

### Boolean
Booleans are a data type that can only have two values: `True` or `False`. They are often used in conditional statements and logical operations.

```python
bl = False
print(bl)
print(type(bl))
```

```python
bool = True
print(bool)
```

### Sequence Types
Sequence types are data types that can hold multiple values in a specific order. The most common sequence types in Python are lists, tuples, and strings.

```python
name = "Sheikh Rasel Ahmed"
print(name)
print(type(name))
```

### Rules of Variable Names
- Variable names must start with a letter (a-z, A-Z) or an underscore (_).
- The rest of the variable name can contain letters, numbers (0-9), and underscores.
- Variable names are case-sensitive (e.g., `myVariable` and `myvariable` are different).
- Variable names cannot be the same as Python keywords (e.g., `if`, `for`, `while`, etc.).

```python
myvar = "John"
my_var = "John"
_my_var = "John"
myVar = "John"
MYVAR = "John"
myvar2 = "John"
```
