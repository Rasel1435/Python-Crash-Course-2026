## round() function in python

পাইথনে round() একটি অত্যন্ত দরকারী বিল্ট-ইন (Built-in) ফাংশন। এটি মূলত কোনো ভাসমান সংখ্যা বা দশমিক সংখ্যাকে (Floating-point number) তার সবচেয়ে কাছের পূর্ণসংখ্যায় (Integer) বা নির্দিষ্ট দশমিক ঘর পর্যন্ত রূপান্তর (Round off) করতে ব্যবহার করা হয়।

### round() ফাংশনের সিনট্যাক্স (Syntax)

```python
number = 70.555555
ndigits = 2
round(number, ndigits)
```

-   number: যে দশমিক সংখ্যাটিকে আপনি রাউন্ড করতে চান (এটি বাধ্যতামূলক)।

-   ndigits: দশমিকের পর আপনি কত ঘর পর্যন্ত রাখতে চান (এটি ঐচ্ছিক বা Optional)। যদি এটি না দেন, তবে পাইথন সরাসরি একটি পূর্ণসংখ্যা (Integer) রিটার্ন করবে।

### বেসিক ব্যবহার (ndigits ছাড়া)
যখন আমরা দ্বিতীয় প্যারামিটারটি (ndigits) দেব না, তখন পাইথন সাধারণ গাণিতিক নিয়মে সবচেয়ে কাছের পূর্ণসংখ্যায় রূপান্তর করবে:

-   দশমিকের পর .5 এর কম হলে আগের সংখ্যায় নামিয়ে দেবে।
-   দশমিকের পর .5 বা তার বেশি হলে পরের সংখ্যায় বাড়িয়ে দেবে।

```python
print(round(10.2))  # আউটপুট: 10 (যেহেতু .2 এর মান .5 এর কম)
print(round(10.7))  # আউটপুট: 11 (যেহেতু .7 এর মান .5 এর চেয়ে বেশি)
```

### নির্দিষ্ট দশমিক ঘর পর্যন্ত রাউন্ড করা (ndigits সহ)
ধরে নিন, আপনার কাছে একটি বড় দশমিক সংখ্যা আছে, কিন্তু আপনি দশমিকের পর মাত্র ২ ঘর বা ৩ ঘর দেখাতে চান।

```python
pi_value = 3.14159265359

print(round(pi_value, 2))  # আউটপুট: 3.14 (দশমিকের পর ২ ঘর)
print(round(pi_value, 4))  # আউটপুট: 3.1416 (দশমিকের পর ৪ ঘর, শেষের 59 হয়ে গেছে 6)
```

### Advanced Concept
সাধারণ গণিতে আমরা জানি .5 থাকলেই পরের সংখ্যায় চলে যায় (যেমন: 2.5 হয়ে যায় 3)। কিন্তু পাইথনে একটু ভিন্ন নিয়ম কাজ করে, যাকে বলা হয় Bankers' Rounding বা Round half to even।

দশমিকের ঠিক মাঝখানে (যেমন .5) থাকলে পাইথন সবসময় তার সবচেয়ে কাছের জোড় সংখ্যায় (Even Number) রাউন্ড করে।

```python
print(round(2.5))  # আউটপুট: 2  (কারণ ২ এবং ৩ এর মধ্যে ২ হলো জোড় সংখ্যা!)
print(round(3.5))  # আউটপুট: 4  (কারণ ৩ এবং ৪ এর মধ্যে ৪ হলো জোড় সংখ্যা!)
```

### .5 থাকলে (যেমন: 5.5, 2.5, 7.5):

যখনই কোনো সংখ্যার শেষে একদম ঠিক .5 থাকবে, পাইথন তার সবচেয়ে কাছের জোড় সংখ্যাটি (Even number) বেছে নেবে।

-   round(5.5) $\rightarrow$ ৫ এবং ৬ এর মধ্যে জোড় সংখ্যা 6
-   round(2.5) $\rightarrow$ ২ এবং ৩ এর মধ্যে জোড় সংখ্যা 2
-   round(7.5) $\rightarrow$ ৭ এবং ৮ এর মধ্যে জোড় সংখ্যা 8
-   round(4.5) $\rightarrow$ ৪ এবং ৫ এর মধ্যে জোড় সংখ্যা 4

### বাকি সব ক্ষেত্রে (দশমিকের পর .5 এর বেশি বা কম হলে):

বাকি সব ক্ষেত্রে পাইথন আমাদের স্কুলের সাধারণ গণিতের নিয়মই মেনে চলে:

-   .5 এর বেশি হলে (যেমন: .৬, .৭, .৮, .৯) $\rightarrow$ আগের সংখ্যার সাথে ১ যোগ করবে (1+)
-   round(5.6) = 6
-   round(2.7) = 3

-   .5 এর কম হলে (যেমন: .১, .২, .৩, .৪) $\rightarrow$ দশমিকের অংশটুকু বাদ দিয়ে আগের সংখ্যাটিই রেখে দেবে (কোনো কিছু যোগ হবে না)।
-   round(5.4) = 5
-   round(2.3) = 2
