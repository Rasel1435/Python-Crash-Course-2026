## ভ্যারিয়েবল (Variable) কী?
সহজ কথায়, Variable হলো একটি পাত্র বা কন্টেইনার (Container) যেখানে আমরা কোনো তথ্য বা ডাটা জমা রাখি।

বাস্তব জীবনের উদাহরণ দিলে— ধরুন আপনার কাছে একটি বক্স আছে যার ওপর আপনি লেবেল লাগিয়েছেন "চিনি"। এখন ওই বক্সের ভেতর আপনি চিনি রাখলেন। এখানে "চিনি" লেখা লেবেলটি হলো ভ্যারিয়েবলের নাম, আর ভেতরের চিনিটুকু হলো ডাটা বা ভ্যালু।

## ভ্যারিয়েবল কীভাবে কাজ করে?
ভ্যারিয়েবল হলো একটি নাম যা একটি ডাটা বা মানকে নির্দেশ করে। যখন আপনি একটি ভ্যারিয়েবল তৈরি করেন এবং তাকে একটি মান দেন, তখন সেই মানটি মেমোরিতে সংরক্ষিত হয় এবং আপনি সেই ভ্যারিয়েবলের নাম ব্যবহার করে সেই মানটি অ্যাক্সেস করতে পারেন।

যখন আপনি কোডিংয়ে লেখেন `x = 10`, তখন পাইথন তিনটি কাজ করে:</br>
সে মেমোরিতে (RAM) একটি জায়গা খোঁজে।</br>
সেখানে `10` সংখ্যাটি রাখে।</br>
ওই মেমোরি লোকেশনটির একটি নাম দেয় `x`, যাতে আপনি পরে এই নাম দিয়ে ডাটাটি খুঁজে পান।

## কম্পিউটারে এটি কীভাবে সেট হয়? (Memory Mapping)
কম্পিউটারে যখন আপনি একটি ভ্যারিয়েবল তৈরি করেন, তখন সেই ভ্যারিয়েবলের জন্য মেমোরিতে একটি নির্দিষ্ট স্থান বরাদ্দ করা হয়। এই স্থানটি হলো একটি ঠিকানা (Address) যা কম্পিউটার ব্যবহার করে সেই ভ্যারিয়েবলের মানটি খুঁজে পেতে।

যখন আপনি `x = 10` লিখেন, তখন কম্পিউটার মেমোরিতে একটি স্থান খুঁজে বের করে যেখানে `10` মানটি সংরক্ষিত হবে এবং সেই স্থানটির একটি ঠিকানা তৈরি করে `x` নামের ভ্যারিয়েবলের সাথে যুক্ত করে। এখন আপনি যখন `x` ব্যবহার করবেন, কম্পিউটার সেই ঠিকানাটি খুঁজে বের করে এবং সেখানে সংরক্ষিত মানটি (যা `10`) আপনাকে প্রদান করবে।

কম্পিউটার আসলে কোনো নাম বোঝে না, সে শুধু বোঝে মেমোরি অ্যাড্রেস (যেমন: `0x7ffee`)। ভ্যারিয়েবল আমাদের এবং কম্পিউটারের মধ্যে একটি সেতু হিসেবে কাজ করে। আমরা নাম (`x`) ব্যবহার করি, আর কম্পিউটার পর্দার আড়ালে সেই নামের বিপরীতে থাকা মেমোরি অ্যাড্রেস থেকে ডাটা তুলে আনে।

## পাইথনে সব ভ্যারিয়েবলকে কেন 'Object' বলা হয়?
পাইথনে সব কিছুই একটি অবজেক্ট (Object)। যখন আপনি একটি ভ্যারিয়েবল তৈরি করেন, তখন আপনি আসলে একটি অবজেক্ট তৈরি করছেন। এই অবজেক্টের একটি টাইপ (Type) থাকে, যা নির্ধারণ করে সেই অবজেক্টটি কী ধরনের ডাটা ধারণ করে।

উদাহরণস্বরূপ, যখন আপনি `x = 10` লিখেন, তখন `10` একটি ইন্টিজার (Integer) অবজেক্ট তৈরি করে এবং `x` সেই অবজেক্টের একটি রেফারেন্স (Reference) হিসেবে কাজ করে। তাই পাইথনে সব ভ্যারিয়েবলকে অবজেক্ট বলা হয়, কারণ তারা সবই একটি অবজেক্টের রেফারেন্স।

এটি পাইথনের একটি ইউনিক বৈশিষ্ট্য। অন্যান্য ল্যাঙ্গুয়েজে (যেমন C বা Java) ভ্যারিয়েবল শুধু একটি ডাটা রাখে। কিন্তু পাইথনে প্রতিটি ডাটা হলো একটি Object।

-   যখন আপনি `x = 5` লেখেন, ৫ শুধু একটি সংখ্যা নয়। এটি পাইথনের একটি `int` ক্লাসের অবজেক্ট।

-   এই অবজেক্টের মধ্যে শুধু ভ্যালু থাকে না, তার সাথে আরও অনেক মেটা-ডাটা থাকে (যেমন তার টাইপ কী, তার আইডি কী ইত্যাদি)। এজন্যই পাইথনকে বলা হয় "Pure Object Oriented Language"।

## অ্যাসাইন অপারেটর (`=`) কীভাবে ভ্যালু অ্যালোকেট করে?
অ্যাসাইন অপারেটর (`=`) একটি ভ্যারিয়েবলের নামের সাথে একটি মান (Value) যুক্ত করে। যখন আপনি `x = 10` লিখেন, তখন `10` মানটি মেমোরিতে সংরক্ষিত হয় এবং `x` নামের ভ্যারিয়েবলটি সেই মানটির রেফারেন্স হিসেবে কাজ করে।

অ্যাসাইন অপারেটরটি একটি স্টেটমেন্ট (Statement) যা একটি ভ্যারিয়েবলের নামের সাথে একটি মানকে যুক্ত করে। এটি একটি এক্সপ্রেশন (Expression) নয়, কারণ এটি কোনো মান রিটার্ন করে না। এটি শুধু একটি অ্যাকশন সম্পাদন করে, যা হলো একটি ভ্যারিয়েবলের নামের সাথে একটি মানকে যুক্ত করা।

পাইথনে `=` চিহ্নটি গাণিতিক সমান নয়, এটি হলো Assignment Operator। এটি ডান পাশের ভ্যালুটিকে বাম পাশের ভ্যারিয়েবল নামের সাথে "লিঙ্ক" বা "বাইন্ড" (Bind) করে দেয়।

কাজের প্রক্রিয়া:

-   প্রথমে ডান পাশের ভ্যালু বা এক্সপ্রেশনটি ক্যালকুলেট হয়।
-   এরপর পাইথন মেমোরিতে সেই ভ্যালুটির জন্য একটি অবজেক্ট তৈরি করে।
-   সবশেষে বাম পাশের নামটি সেই অবজেক্টের দিকে পয়েন্ট করে।

## RAM-এ কীভাবে ভ্যারিয়েবল অ্যাড্রেস তৈরি হয়?
RAM-এ যখন একটি ভ্যারিয়েবল তৈরি করা হয়, তখন কম্পিউটার মেমোরিতে একটি নির্দিষ্ট স্থান বরাদ্দ করে। এই স্থানটি হলো একটি ঠিকানা (Address) যা কম্পিউটার ব্যবহার করে সেই ভ্যারিয়েবলের মানটি খুঁজে পেতে।

এটি বোঝার জন্য আমাদের পাইথনের মেমোরি ম্যানেজমেন্ট বুঝতে হবে।</br>
-   Object Creation: যখন আপনি কোনো ভ্যালু দেন, RAM-এর একটি অংশ যাকে বলা হয় Heap Memory, সেখানে ওই ডাটার জন্য জায়গা বরাদ্দ করা হয়।

-   Reference: ভ্যারিয়েবল নামটি নিজে ডাটা ধরে রাখে না, সে শুধু ওই মেমোরি অ্যাড্রেসের রেফারেন্স (Reference) বা ঠিকানা ধরে রাখে।

আপনি চাইলে পাইথনে `id()` ফাংশন ব্যবহার করে এই অ্যাড্রেসটি দেখতে পারেন:

```python
x = 10
print(id(x)) # এটি আপনাকে RAM-এ x এর মেমোরি অ্যাড্রেস দেখাবে
```

```python
a = 5
b = 5
print(id(a))
print(id(b))
```

এখানে দেখবেন `a` এবং `b` দুজনের আইডি একই! কারণ পাইথন মেমোরি বাঁচাতে একই ভ্যালুর জন্য আলাদা অ্যাড্রেস তৈরি না করে একই অবজেক্টে পয়েন্ট করে। একে বলা হয় Object Interning।

```python
# Variables and Assignment
x = 10
print(x)

# Checking RAM Memory Address
print(id(x))

# Object Interning Example (Memory Optimization)
a = 5
b = 5
print(id(a) == id(b)) # Returns True!
```
