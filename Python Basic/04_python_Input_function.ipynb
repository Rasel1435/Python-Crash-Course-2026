## Input Function()

পাইথনে input() ফাংশনটি মূলত ব্যবহারকারী বা ইউজারের কাছ থেকে সরাসরি কিবোর্ডের মাধ্যমে ডেটা বা ইনপুট নেওয়ার জন্য ব্যবহার করা হয়। প্রোগ্রামকে ইন্টারেক্টিভ (Interactive) বা ব্যবহারকারীর সাথে যোগাযোগক্ষম করে তুলতে এই ফাংশনটির ভূমিকা অপরিসীম।

যেকোনো প্রোগ্রামে ইউজারের নাম, বয়স, ইমেইল বা পাসওয়ার্ডের মতো ব্যক্তিগত তথ্য সাময়িকভাবে সংগ্রহ করে তা ভ্যারিয়েবলে জমা রাখতে এটি ব্যবহার করা হয়।

```python
name = input("Enter your name: ")
print("Welcome,", name)
```

## TypeCasting with Input Function()
পাইথনে input() ফাংশন দিয়ে ইউজার যা-ই ইনপুট দিক না কেন (তা কোনো লেখা হোক বা সংখ্যা), পাইথন সেটিকে সবসময় String (টেক্সট) হিসেবে গ্রহণ করে। তাই গাণিতিক কাজ করার সময় এটিকে int() বা float() দিয়ে কনভার্ট বা টাইপ কাস্ট করে নিতে হয়।

### str

```python
string = input("Enter The String Value: ")
print(string)
print(type(string))
```

### int

```python
integer = int(input("Enter The Integer Value: "))
print(integer)
print(type(integer))
```

### float

```python
floting = float(input("Enter The Floating Value: "))
print(floting)
print(type(floting))
```
