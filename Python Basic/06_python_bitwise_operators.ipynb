## Python Bitwise Operators
Bitwise operators are used to compare (binary) numbers. In Python, there are several bitwise operators that you can use:
- `&` (AND): Sets each bit to 1 if both bits are 1
- `|` (OR): Sets each bit to 1 if one of the bits is 1
- `^` (XOR): Sets each bit to 1 if only one of the bits is 1
- `~` (NOT): Flips all the bits
- `<<` (Left shift): Shifts bits to the left and fills with zeros
- `>>` (Right shift): Shifts bits to the right and fills with zeros




### `&` (AND) Operator
Sets each bit to 1 if both bits are 1

AND অপারেটর তখনই 1 দেয়, যখন দুটি বিট-ই 1 হয়। অন্য সব ক্ষেত্রে 0 হয়।

#### 3 Bits Formula

````
      4  2  1  <-- (বিট পজিশন বা হেডার)
________________
5 =   1  0  1       (4 + 1 = 5)
3 =   0  1  1       (2 + 1 = 3)
________________
R =   0  0  1  (শুধুমাত্র 1 এর নিচে দুটি 1 আছে, তাই 1 এর মানটি নেওয়া হলো)

হিসাব (Result):
এখানে ৪ এবং ২ এর নিচে 0 এসেছে এবং ১ এর নিচে 1 এসেছে,
তাহলে ডেসিমেল মান হবে: 0 + 0 + 1 = 1

আউটপুট = 1
````

`````python
a = 5 # In binary: 0101
b = 3 #In binary: 0011
result = a & b #In binary: 0001
print(result)
`````

### `|` (OR) Operator
Sets each bit to 1 if one of the bits is 1

OR অপারেটরের ক্ষেত্রে যেকোনো একটি বিট 1 হলেই আউটপুট 1 হয়। দুটি বিট-ই 0 হলে আউটপুট 0 হয়।

#### 3 Bits Formula

````
      4  2  1  <-- (বিট পজিশন বা হেডার)
________________
5 =   1  0  1       (4 + 1 = 5)
3 =   0  1  1       (2 + 1 = 3)
________________
R =   1  1  1  (যেকোনো একটি বিট 1 হলে আউটপুট 1 হয়, তাই সব বিট-ই 1 হয়েছে)

হিসাব (Result):
এখানে ৪, ২, এবং ১ এর নিচে ১ এসেছে
তাহলে ডেসিমেল মান হবে: 4 + 2 + 1 = 7

আউটপুট = 7
````

`````python
a | b
`````

### `^` (XOR) Operator
Sets each bit to 1 if only one of the bits is 1

XOR অপারেটর তখনই 1 দেয়, যখন দুটি বিট-ই আলাদা হয়। অর্থাৎ, একটি বিট 1 এবং অন্যটি 0 হলে আউটপুট 1 হয়। যদি দুটি বিট-ই একই হয় (দুটি 0 বা দুটি 1), তাহলে আউটপুট 0 হয়।
#### 3 Bits Formula

````
      4  2  1  <-- (বিট পজিশন বা হেডার)
________________
5 =   1  0  1       (4 + 1 = 5)
3 =   0  1  1       (2 + 1 = 3)
________________
R =   1  1  0  (শুধুমাত্র 1 এর নিচে দুটি বিট-ই আলাদা, তাই 1 এবং 2 এর মান নেওয়া হলো)

হিসাব (Result):
এখানে ৪ এবং ২ এর নিচে ১ এসেছে এবং ১ এর নিচে 0 এসেছে
তাহলে ডেসিমেল মান হবে: 4 + 2 + 0 = 6

আউটপুট = 6
````


`````python
a ^ b
`````

### `~` (NOT) Operator
Flips all the bits

NOT অপারেটর একটি বিটের মান উল্টে দেয়। অর্থাৎ, 1 কে 0 করে এবং 0 কে 1 করে।

#### 4 Bits Explanation with Sign Bit
কম্পিউটার পজিটিভ (+) এবং নেগেটিভ (-) সংখ্যা বোঝানোর জন্য একদম বামের বিটটিকে সাইন বিট (Sign Bit) হিসেবে ব্যবহার করে।

৪-বিটের ক্ষেত্রে আমাদের হেডার বা স্থানিক মান হবে: -৮, ৪, ২, ১।

```
   -8  4  2  1  <-- (বিট পজিশন বা হেডার)
________________
5 = 0  1  0  1  (৪ + ১ = ৫। সাইন বিট ০, তাই এটি পজিটিভ)
________________
R = 1  0  1  0  (NOT অপারেটর সব বিট উল্টে দেওয়ায় ৫ হয়ে গেল ১০১০)
```

কম্পিউটার কীভাবে এই নেগেটিভ মানটি হিসাব করে? (২'স কমপ্লিমেন্ট নিয়ম)

১০১০ এর ডেসিমেল মান বের করার সহজ নিয়ম: হেডারের নিচে যেখানে ১ আছে সেগুলো যোগ করা।

```
   এখানে -৮ এর নিচে ১ আছে এবং ২ এর নিচে ১ আছে।
   অতএব, -8 + 2 = -6
   পাইথনে print(~5) দিলে আউটপুট আসবে -6।
```
#### শর্টকাট সূত্র (Formula)
NOT অপারেটরের জন্য একটি সহজ সূত্র হলো: `~x = -(x + 1)`
```
5 = -(5 + 1) = -6
10 = -(10 + 1) = -11
~(-5) = -(-5 + 1) = -(-4) = 4
```

`````python
~a
`````

### `<<` (Left shift) Operator

Shifts bits to the left and fills with zeros

এই অপারেটরটি বাইনারি বিটগুলোকে বাম দিকে নির্দিষ্ট সংখ্যক ঘর সরিয়ে দেয় এবং ডানদিকের খালি ঘরগুলো 0 দিয়ে পূরণ করে। এর ফলে সংখ্যাটির মান দ্বিগুণ হারে বাড়ে।
#### 4 Bits Formula

````
      8  4  2  1  <-- (বিট পজিশন বা হেডার)
________________
5 =   1  0  1       (4 + 1 = 5)________________
R =   1  0  1  0  (বাম দিকে 1 ঘর সরানো হয়েছে, তাই 4 এর নিচে 1, 2 এর নিচে 0, 1 এর নিচে 1 এবং নতুন ঘর (0) এর নিচে 0 এসেছে)
````
#### শর্টকাট সূত্র
````
x << n = x * (2^n)
````
এখানে x = সংখ্যাটি (৫) এবং n = কত ঘর সরাতে হবে (১)।

সূত্র অনুযায়ী: 5 * 2^1 = 5 * 2 = 10।

`````python
a << b
`````

### `>>` (Right shift) Operator
Shifts bits to the right and fills with zeros

এই অপারেটরটি বাইনারি বিটগুলোকে ডান দিকে নির্দিষ্ট সংখ্যক ঘর সরিয়ে দেয়। এর ফলে একদম ডানদিকের বিটগুলো মেমরি থেকে বাদ (Drop) পড়ে যায় এবং বামদিকের খালি ঘরগুলো 0 দিয়ে পূরণ হয়। রাইট শিফটের ফলে সংখ্যাটির মান অর্ধেক হারে কমে।

#### 4 Bits Formula

````
    8  4  2  1  <-- (বিট পজিশন বা হেডার)
________________
5 =    0  1  0  1       (৪ + ১ = ৫)
________________
R = 0  0  1  0  [1]  (ডান দিকে ১ ঘর সরানো হয়েছে। একদম ডানের ১ নম্বর বিটটি বাদ পড়ে গেছে এবং বামে একটি নতুন ০ বসেছে)
````
ডানদিকের বিটটি বাদ পড়ার পর নতুন বাইনারিটি দাঁড়াল 0010 যার ডেসিমেল মান ২।

#### শর্টকাট সূত্র
````
x >> n = x // (2^n)
````
(এখানে // বা Floor Division মানে হলো ভাগ করার পর দশমিকের পরের অংশ বাদ দিয়ে শুধু পূর্ণসংখ্যা নেওয়া)

এখানে x = সংখ্যাটি (৫) এবং n = কত ঘর সরাতে হবে (১)।

সূত্র অনুযায়ী: 5 // 2^1 = 5 // 2 = 2।

(৫ কে ২ দিয়ে ভাগ করলে ২.৫ হয়, কিন্তু দশমিক বাদ দিয়ে উত্তর হবে ২)

`````python
a >> b
`````

`````python
a = 5  # বাইনারি: 101
b = 3  # বাইনারি: 011

print(a & b)   # আউটপুট: 1  (AND)
print(a | b)   # আউটপুট: 7  (OR)
print(a ^ b)   # আউটপুট: 6  (XOR)
print(~a)      # আউটপুট: -6 (NOT)
print(a << 1)  # আউটপুট: 10 (Left Shift)
print(a >> 1)  # আউটপুট: 2  (Right Shift)
`````
