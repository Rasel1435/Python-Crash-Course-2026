## Python Print() Function

The `print()` function in Python is used to output data to the standard output device (usually the console). It can take multiple arguments and will convert them to strings before printing. The `print()` function also has several optional parameters that allow you to customize the output.

````python
print("My name is Sheikh Rasel")
````

## Case Sensitivity in Python
python is a case-sensitive language, which means that `print` and `Print` are considered different identifiers. The correct usage is `print()` with a lowercase 'p'.

````python
Print('My name is Sheikh Rasel')
````

if you try to use `Print()` with an uppercase 'P', you will get a `NameError` because Python does not recognize it as a valid function. Always ensure that you use the correct case when calling functions in Python. it's a common mistake for beginners to forget that Python is case-sensitive, so it's important to pay attention to the case of your function names. it's a bultin function, so it should always be written in lowercase. we cant change the case of built-in functions, so we must use `print()` with a lowercase 'p' to avoid errors.

## Double Quotes and Single Quotes in Python
In Python, you can use either double quotes (`"`) or single quotes (`'`) to define string literals. Both are valid and functionally equivalent, but they allow you to include certain characters without needing to escape them.

````python
print("My Name is 'Sheikh' Rasel")
````

but if you want to include a double quote inside a string, you can use single quotes to define the string, and vice versa. For example:

```python
print('My Name is "Sheikh" Rasel')
```
we cant use double quotes to define the string if we want to include a double quote inside it without escaping, and we cant use single quotes to define the string if we want to include a single quote inside it without escaping.

````python
print("My Name is "Sheikh" Rasel")
````

````python
print('My Name is 'Sheikh' Rasel')
````

## Multi-line Strings in Python
Multiple lines of text can be printed using triple quotes (`'''` or `"""`). This allows you to create multi-line strings without needing to use newline characters (`\n`).

````python
print("""This is a multi-line string.
It can span multiple lines without needing to use newline characters.""")
````

````python
print('''This is a multi-line string.
It can span multiple lines without needing to use newline characters.''')
````

or we can use (`\n`) to create a new line in a string defined with single or double quotes:

````python
print("This is a multi-line string.\nIt can span multiple lines using newline characters.")
````

````python
print("My name is Sheikh Rasel. \nMy Age is 24. \nI love My Country")
````

## String Concatenation in Python
You can concatenate strings in Python using the `+` operator. This allows you to combine multiple strings into one. For example:

````python
print("My name is Sheikh Rasel"+" "+"My Age is 24"+" "+"HI")
````
