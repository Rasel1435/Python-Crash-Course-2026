## Python f-string (Formatted String Literal)
পাইথনে f-string (Formatted String Literal) হলো স্ট্রিং ফরম্যাট করার বা কোনো স্ট্রিংয়ের ভেতরে ভ্যারিয়েবল ও এক্সপ্রেশনের মান দেখানোর সবচেয়ে আধুনিক, সহজ এবং দ্রুততম (Fastest) নিয়ম। পাইথন ৩.৬ সংস্করণে এটি প্রথম আনা হয়।

#### বেসিক ধারণা (Basic Concept)
আগে পাইথনে স্ট্রিংয়ের সাথে ভ্যারিয়েবল জোড়া দেওয়ার জন্য + অপারেটর বা .format() মেথড ব্যবহার করা হতো, যা বেশ জটিল ছিল। f-string আসার পর সেই ঝামেলা একদম শেষ।

নিয়ম: যেকোনো স্ট্রিং লেখার শুরুতে শুধু একটি f অথবা F লিখে দিতে হবে। এরপর স্ট্রিংয়ের ভেতরে যেখানে ভ্যারিয়েবল বসাতে চান, সেখানে সেকেন্ড ব্র্যাকেট বা কার্লি ব্রেস {} ব্যবহার করতে হবে।

```python
name = "Rasel"
age = 25

# f-string ব্যবহার করে:
print(f"My name is {name} and I am {age} years old.")
```

#### f-string এর ভেতর গাণিতিক হিসাব (Expressions)
f-string এর সবচেয়ে বড় সুবিধা হলো, আপনি {} ব্র্যাকেটের ভেতরে শুধু ভ্যারিয়েবল নয়, সরাসরি যেকোনো পাইথন কোড বা গাণিতিক হিসাবও করে ফেলতে পারেন।

```python
a = 10
b = 20

print(f"The sum of a and b is {a + b}")
print(f"Upper case: {'hello'.upper()}")
```

#### দশমিকের পর ঘর নির্ধারণ করা (Floating-point Formatting)
ধরুন, কোনো হিসাবের পর দশমিকের পর অনেকগুলো সংখ্যা আসছে (যেমন: 28.36754), কিন্তু আপনি শুধু দশমিকের পর ২ ঘর দেখাতে চান। f-string দিয়ে এটি ১ লাইনে করা যায়:

```python
bmi_value = 28.36754

# :.2f মানে হলো দশমিকের পর (f = float) মাত্র ২ ঘর দেখাও
print(f"Your precise BMI is {bmi_value:.2f}")  
```

#### কেন f-string ব্যবহার করব?
এটি দেখতে পরিষ্কার (Readable), কোড লিখতে সময় কম লাগে এবং পাইথনের পুরনো নিয়মগুলোর (% বা .format()) চেয়ে ব্যাক-এন্ডে অনেক দ্রুত রান করে।
