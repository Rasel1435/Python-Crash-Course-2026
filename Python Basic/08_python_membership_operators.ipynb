## Membership Operators
কোনো একটি নির্দিষ্ট উপাদান (Sequence বা Element) অন্য কোনো বড় ডাটা কালেকশনের (যেমন: String, List, Tuple, বা Dictionary) মধ্যে উপস্থিত আছে কি না, তা খুঁজে বের করার জন্য Membership Operators ব্যবহার করা হয়।

এর আউটপুট সবসময় একটি বুলিয়ান মান— অর্থাৎ True অথবা False আসে।

পাইথনে মেম্বারশিপ অপারেটর দুটি:
-   in
-   not in

#### in Operator (ভেতরে আছে কি?)

যদি কোনো নির্দিষ্ট উপাদানটি ওই ডাটা স্ট্রাকচার বা সিকোয়েন্সের মধ্যে থাকে, তবে এটি True রিটার্ন করে। আর যদি না থাকে, তবে False রিটার্ন করে।

```python
fruits = ["apple", "banana", "mango"]

print("banana" in fruits)  # আউটপুট: True (কারণ banana লিস্টে আছে)
print("orange" in fruits)  # আউটপুট: False (কারণ orange লিস্টে নেই)
```

#### not in Operator (ভেতরে নেই তো?)
এটি in অপারেটরের ঠিক উল্টো কাজ করে। যদি নির্দিষ্ট উপাদানটি ওই ডাটা কালেকশনের মধ্যে না থাকে, তবে এটি True রিটার্ন করে। আর যদি উপাদানটি থেকে যায়, তবে False রিটার্ন করে।

```python
blocked_users = ["user123", "spammer99"]

print("rasel42" not in blocked_users)  # আউটপুট: True (কারণ সে ব্লকলিস্টে নেই)
print("user123" not in blocked_users)  # আউটপুট: False (কারণ সে ব্লকলিস্টে আছে)
```

#### String (স্ট্রিং)-এর ক্ষেত্রে সাব-স্ট্রিং চেক:
স্ট্রিংয়ের ক্ষেত্রে এটি শুধু সিঙ্গেল ক্যারেক্টার নয়, বরং সাব-স্ট্রিং (Sub-string) বা পুরো একটি শব্দও চেক করতে পারে। তবে এটি Case-sensitive (বড় হাতের ও ছোট হাতের অক্ষর আলাদা বিবেচনা করে)।

```python
sentence = "Welcome to Python Crash Course 2026"

print("Python" in sentence)  # আউটপুট: True
print("python" in sentence)  # আউটপুট: False (ছোট হাতের 'p' হওয়ার কারণে)
```

#### Dictionary (ডিকশনারি)-এর ক্ষেত্রে সিক্রেট ট্রিক:
এটি ইন্টারভিউতে খুব বেশি জিজ্ঞেস করা হয়! ডিকশনারির ওপর সরাসরি মেম্বারশিপ অপারেটর চালালে তা সবসময় শুধুমাত্র Keys (চাবি) গুলোকে খোঁজে, ভেতরের Values (মান) খোঁজে না।

```python
user_info = {
    "name": "Rasel",
    "role": "Instructor"
    }

print("name" in user_info)  # আউটপুট: True (কারণ 'name' একটি Key)
print("Rasel" in user_info) # আউটপুট: False (কারণ 'Rasel' একটি Value, Key নয়!)
```

.values() মেথড ব্যবহার করে ডিকশনারির ভ্যালুগুলোকে চেক করা যায়।
